### Configuração Inicial e Modelo Base

```python
# Instala bibliotecas necessárias (caso não estejam disponíveis)
!pip install -q transformers torch

from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch

# Configura o dispositivo (GPU se disponível)
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Carrega o modelo e o tokenizador do GPT-2
model = GPT2LMHeadModel.from_pretrained('gpt2').to(device)
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model.eval()

# Texto de entrada original do artigo
text = "I have a dream"
input_ids = tokenizer.encode(text, return_tensors='pt').to(device)

print(f"Ambiente configurado com sucesso! Dispositivo em uso: {device.upper()}")
```

### Greedy Search

```python
# Geração utilizando Greedy Search (padrão quando do_sample=False)
outputs = model.generate(
    input_ids,
    max_length=len(input_ids.squeeze()) + 5,
    do_sample=False
)

generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print("=== GREEDY SEARCH ===")
print(f"Texto gerado: {generated_text}")
```

### Variação de Temperatura

```python
# 1. Temperatura Baixa (Conservador / Mais previsível)
outputs_low = model.generate(
    input_ids,
    max_length=len(input_ids.squeeze()) + 5,
    do_sample=True,
    temperature=0.2
)
print("--- Temperatura Baixa (0.2) ---")
print(tokenizer.decode(outputs_low[0], skip_special_tokens=True))

# 2. Temperatura Alta (Criativo / Mais diversificado)
outputs_high = model.generate(
    input_ids,
    max_length=len(input_ids.squeeze()) + 5,
    do_sample=True,
    temperature=1.2
)
print("\n--- Temperatura Alta (1.2) ---")
print(tokenizer.decode(outputs_high[0], skip_special_tokens=True))
```

### Top-K Sampling

```python
# Top-K Sampling (ex: considerando apenas os 20 tokens mais prováveis)
outputs_topk = model.generate(
    input_ids,
    max_length=len(input_ids.squeeze()) + 5,
    do_sample=True,
    temperature=0.7,
    top_k=20
)

generated_text = tokenizer.decode(outputs_topk[0], skip_special_tokens=True)
print("=== TOP-K SAMPLING (k=20) ===")
print(f"Texto gerado: {generated_text}")
```

### Top-P

```python
# Nucleus / Top-P Sampling (ex: mantendo apenas o núcleo de 90% de probabilidade cumulativa)
outputs_topp = model.generate(
    input_ids,
    max_length=len(input_ids.squeeze()) + 5,
    do_sample=True,
    temperature=0.7,
    top_p=0.9
)

generated_text = tokenizer.decode(outputs_topp[0], skip_special_tokens=True)
print("=== TOP-P / NUCLEUS SAMPLING (p=0.9) ===")
print(f"Texto gerado: {generated_text}")
```
