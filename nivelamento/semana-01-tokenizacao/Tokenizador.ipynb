```python
# Instalação da biblioteca oficial da OpenAI
!pip install tiktoken -q

import tiktoken
import matplotlib.pyplot as plt

# Carregamento do tokenizador usado pelo GPT-4, que será nosso modelo de teste
enc = tiktoken.get_encoding("cl100k_base")

def comparar_idiomas(texto_en, texto_pt):
    # Tradutor de texto em tokens
    tokens_en = enc.encode(texto_en)
    tokens_pt = enc.encode(texto_pt)

    # Calculo da diferença percentual
    diferenca = ((len(tokens_pt) / len(tokens_en)) - 1) * 100

    # Resultados
    print("RESULTADOS DA TOKENIZAÇÃO")
    print(f"Inglês: {len(tokens_en)} tokens")
    print(f"Português: {len(tokens_pt)} tokens")
    print(f"O português exigiu {diferenca:.1f}% mais tokens (e custaria {diferenca:.1f}% a mais na API)\n")

    # Geração do gráfico visual
    idiomas = ['Inglês', 'Português']
    quantidades = [len(tokens_en), len(tokens_pt)]

    plt.figure(figsize=(6, 4))
    plt.bar(idiomas, quantidades, color=['#FFB7CE', '#FF1D8D'])
    plt.title('Comparação de Custo e Fragmentação (Tokens)')
    plt.ylabel('Quantidade de Tokens')
    plt.show()

# 3. TEXTO PARA DEMONSTRAÇÃO
texto_ingles = "hey"
texto_portugues = "olá"

# 4. Execução da função
comparar_idiomas(texto_ingles, texto_portugues)
```
